```python
user_input = input("Enter numbers separated by spaces: ")

numbers = [int(x) for x in user_input.split()]

largest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num

print("The largest number in the list is:", largest)
```

```python
text = input("Enter a string: ")

word_to_delete = input("Enter the word you want to delete: ")

words = text.split()
new_words = [word for word in words if word != word_to_delete]
new_text = " ".join(new_words)

print("String after deleting the word:", new_text)
```

```python
months = {
    '01': 'January', '02': 'February', '03': 'March', '04': 'April',
    '05': 'May', '06': 'June', '07': 'July', '08': 'August',
    '09': 'September', '10': 'October', '11': 'November', '12': 'December'
}

date_input = input("Enter a date (mm/dd/yyyy): ")

month, day, year = date_input.split('/')

month_name = months[month]
formatted_date = f"{month_name} {int(day)}, {year}"

print("Formatted date:", formatted_date)
```

```python
m = int(input("Enter number of rows (m): "))
n = int(input("Enter number of columns (n): "))

matrix = []

print("Enter the elements row by row:")
for i in range(m):
    row = list(map(int, input(f"Enter elements of row {i+1}, separated by spaces: ").split()))
    while len(row) != n:
        print(f"Please enter exactly {n} elements.")
        row = list(map(int, input(f"Enter elements of row {i+1} again: ").split()))
    matrix.append(row)

for i in range(m):
    row_sum = sum(matrix[i])
    print(f"Sum of row {i+1} = {row_sum}")
```

```python
n = int(input("Enter number of rows (n): "))
m = int(input("Enter number of columns (m): "))

print("Enter elements of first matrix:")
matrix1 = []
for i in range(n):
    row = list(map(int, input(f"Enter elements of row {i+1} (separated by spaces): ").split()))
    while len(row) != m:
        print(f"Please enter exactly {m} elements.")
        row = list(map(int, input(f"Enter elements of row {i+1} again: ").split()))
    matrix1.append(row)

print("Enter elements of second matrix:")
matrix2 = []
for i in range(n):
    row = list(map(int, input(f"Enter elements of row {i+1} (separated by spaces): ").split()))
    while len(row) != m:
        print(f"Please enter exactly {m} elements.")
        row = list(map(int, input(f"Enter elements of row {i+1} again: ").split()))
    matrix2.append(row)

result = []
for i in range(n):
    row = []
    for j in range(m):
        row.append(matrix1[i][j] + matrix2[i][j])
    result.append(row)

print("\nResultant Matrix after Addition:")
for row in result:
    print(row)
```

```python
n = int(input("Enter number of rows of first matrix (n): "))
m = int(input("Enter number of columns of first matrix / rows of second matrix (m): "))
p = int(input("Enter number of columns of second matrix (p): "))

print("\nEnter elements of first matrix (A):")
A = []
for i in range(n):
    row = list(map(int, input(f"Enter elements of row {i+1} (separated by spaces): ").split()))
    while len(row) != m:
        print(f"Please enter exactly {m} elements.")
        row = list(map(int, input(f"Enter elements of row {i+1} again: ").split()))
    A.append(row)

print("\nEnter elements of second matrix (B):")
B = []
for i in range(m):
    row = list(map(int, input(f"Enter elements of row {i+1} (separated by spaces): ").split()))
    while len(row) != p:
        print(f"Please enter exactly {p} elements.")
        row = list(map(int, input(f"Enter elements of row {i+1} again: ").split()))
    B.append(row)

C = [[0 for _ in range(p)] for _ in range(n)]

for i in range(n):
    for j in range(p):
        for k in range(m):
            C[i][j] += A[i][k] * B[k][j]

print("\nResultant Matrix (A × B):")
for row in C:
    print(" ".join(map(str, row)))
```
